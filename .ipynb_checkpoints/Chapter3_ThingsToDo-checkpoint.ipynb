```python
60 * 60
```

There are 3600 seconds in an hour.

```python
seconds_per_hour = 60 * 60
seconds_per_hour
```

```python
seconds_per_hour * 24
```

There are 86,400 seconds in a day.

```python
seconds_per_day = seconds_per_hour * 24
seconds_per_day
```

```python
seconds_per_day / seconds_per_hour
```

```python
seconds_per_day // seconds_per_hour
```

Yes the result agrees with the floating-point value, aside from the .0.
